```python
# Importando Bibliotecas de conexão
import mysql.connector
```

```python
# Acessando o banco de dados
bd = mysql.connector.connect(
    host = "localhost",
    user = "root",
    password = "raphael",
    database = "World_Cup_26"
)

mycursor = bd.cursor()

mycursor.execute("SELECT * FROM seleções")

for x in mycursor:
  print(x)
```

```python
grupo_a = []
grupo_b = []
grupo_c = []
grupo_d = []
grupo_e = []
grupo_f = []
grupo_g = []
grupo_h = []
grupo_i = []
grupo_j = []
grupo_k = []
grupo_l = []
```

```python
# O loop começa aqui
for x in mycursor:
    # Criamos uma variável temporária para facilitar a leitura e limpar o dado
    grupo_atual = x[5].strip().upper()
    
    if grupo_atual == "A":
        grupo_a.append(x)
    elif grupo_atual == "B":
        grupo_b.append(x)
    elif grupo_atual == "C":
        grupo_c.append(x)
    elif grupo_atual == "D":
        grupo_d.append(x)
    elif grupo_atual == "E":
        grupo_e.append(x)
    elif grupo_atual == "F":
        grupo_f.append(x)
    elif grupo_atual == "G":
        grupo_g.append(x)
    elif grupo_atual == "H":
        grupo_h.append(x)
    elif grupo_atual == "I":
        grupo_i.append(x)
    elif grupo_atual == "J":
        grupo_j.append(x)
    elif grupo_atual == "K":
        grupo_k.append(x)
    elif grupo_atual == "L":
        grupo_l.append(x)
    else:
        
        print(f"Atenção: Seleção {x[1]} está num grupo fora de A-L: {x[5]}")
```

```python
print("-" * 20)
print(f"Grupo A: {len(grupo_a)} times")
print(f"Grupo B: {len(grupo_b)} times")
print(f"Grupo C: {len(grupo_c)} times")
# ... continue para os outros
print("-" * 20)
```
