```python
lista_df = []

for pais in ajustes['pais']:
    # Barra pesquisar
    pesquisa = browser.find_element("xpath", '//*[@id="search-trigger-button"]/div')
    pesquisa.click()

    # Digitar a seleção
    try:
        digite = browser.find_element("xpath", '//*[@id="search-input"]')
        digite.click()
        digite.send_keys(pais)
        print('Selenium')
    except:
        pyautogui.click(493, 144)
        pyperclip.copy(pais)
        pyautogui.hotkey('ctrl', 'v')
        time.sleep(10)
        # print('pyautogui')

    # Selecionar o pais
    # try:
    #     selecione = browser.find_element('xpath', '/html/body/div[6]/div[2]/div/div[2]/div/div[1]/a/div')
    #     selecione.click()
    #     selecione.click()
    #     print('Selenium')
    # except:
    #     pyautogui.click(441, 279)
    #     time.sleep(2)
    #     # print('pyautogui')

    # Digitar botão de estatisticas
    # try:
    #     estatisticas = browser.find_element("xpath", '//*[@id="horizontally-scrollable«re8»"]/div/div/div/button[2]')
    #     estatisticas.click()
    # except:
    #     pyautogui.click(x=870, y=423)
    #     time.sleep(1)

    # Copiando toda a página
    pyautogui.hotkey('ctrl', 'a')
    time.sleep(0.5)
    pyautogui.hotkey('ctrl', 'c')

    time.sleep(0.5)
    texto = pyperclip.paste()

    # Nome do time
    time_nome = pais

    # Filtrando
    parte = texto.split("Geral")[1].split("Cartões vermelhos")[0]

    # Limpando texto
    linhas = [l.strip() for l in parte.split('\n') if l.strip()]

    # Palavras para remover
    remover = ['Geral', 'Atacando', 'Passes', 'Defendendo', 'Outros']

    # Removendo títulos
    linhas = [l for l in linhas if l not in remover]

    # Criando dicionário
    dados = {}

    for i in range(0, len(linhas)-1, 2):
        chave = linhas[i]
        valor = linhas[i+1]

        dados[chave] = valor

    # Adicionando nome do time
    dados['Time'] = time_nome

    # Transformando em DataFrame com estatísticas em colunas
    df = pd.DataFrame([dados])

    # Colocando Time como primeira coluna
    colunas = ['Time'] + [c for c in df.columns if c != 'Time']
    df = df[colunas]

    # Salvando dataframe na lista
    lista_df.append(df)

    # Exibindo tabela
    print(df)

df_final_ajustes = pd.concat(lista_df, ignore_index=True)

print(df_final_ajustes)
```
